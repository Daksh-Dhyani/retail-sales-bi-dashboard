```python
import pandas as pd

df = pd.read_csv("../data/superstore_clean.csv", parse_dates=["Order Date", "Ship Date"])
print("Rows, columns:", df.shape)

# Check 1: missing values and duplicates
print("Missing values:", df.isna().sum().sum())
print("Duplicate rows:", df.duplicated().sum())
print("Duplicate Row IDs:", df["Row ID"].duplicated().sum())

# Check 2: impossible values
print("Sales <= 0:", (df["Sales"] <= 0).sum())
print("Quantity <= 0:", (df["Quantity"] <= 0).sum())
print("Discount outside 0-1:", ((df["Discount"] < 0) | (df["Discount"] > 1)).sum())
print("Shipped before ordered:", (df["Ship Date"] < df["Order Date"]).sum())
```

```python
# Check 3: totals must match the dashboard
assert round(df["Sales"].sum(), 2) == 2297200.86, "Sales total is wrong"
assert round(df["Profit"].sum(), 2) == 286397.02, "Profit total is wrong"
assert df["Order ID"].nunique() == 5009, "Order count is wrong"
assert round(df["Profit"].sum() / df["Sales"].sum() * 100, 2) == 12.47, "Margin is wrong"
print("All total checks passed")

# Check 4: region table to compare with the dashboard
region = df.groupby("Region").agg(
    Sales=("Sales", "sum"),
    Profit=("Profit", "sum"),
    Orders=("Order ID", "nunique"),
).round(2)
region["Margin %"] = (region["Profit"] / region["Sales"] * 100).round(2)
print(region)
```
